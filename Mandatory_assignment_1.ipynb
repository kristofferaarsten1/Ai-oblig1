Det inneholder 5 unike Item Fat Content-typer:
Low Fat, Regular, low fat, LF og reg.

Ja, det er et problem. De fem verdiene representerer bare to reelle
kategorier som er LF og low fat er skrivevarianter av Low Fat, og reg er en
variant av Regular. Pandas behandler dem som fem forskjellige kategorier,
noe som ville gitt feil ved gruppering, telling og koding senere.

Problemet er håndtert med replace(). Etter oppryddingen er det 2 unike
typer: Low Fat (3253 rader) og Regular (1747 rader). Summene stemmer:
3008 + 183 + 62 = 3253 og 1679 + 68 = 1747, så ingen rader er tapt.

```python
# Antall unike verdier og hvilke de er
print("Antall unike Item Fat Content:", Grc_df["Item_Fat_Content"].nunique())
print("Unike verdier:", Grc_df["Item_Fat_Content"].unique())

# Fordeling per verdi
print("\nAntall rader per type:")
print(Grc_df["Item_Fat_Content"].value_counts())

# LF og low fat er skrivevarianter av Low Fat, reg er variant av Regular.
# Vi standardiserer til to kategorier.
Grc_df["Item_Fat_Content"] = Grc_df["Item_Fat_Content"].replace({
    "LF": "Low Fat",
    "low fat": "Low Fat",
    "reg": "Regular"
})

# Kontroll etter opprydding
print("\nAntall unike etter opprydding:", Grc_df["Item_Fat_Content"].nunique())
print(Grc_df["Item_Fat_Content"].value_counts())
```
